# EV Charging Data Visualization

This notebook covers the exploratory data analysis and visualization section for the EV charging forecasting project.

Objectives:
- summarize the processed datasets used in the project
- convert timestamps from UTC to California local time before building time-based charts
- generate report-ready figures for the final report and presentation slides

Important note: the raw timestamps are stored in UTC. This notebook converts them to California local time with a manual Pacific DST rule so it can run in browser-based Jupyter environments that do not ship with `tzdata`.


```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display

ipython = get_ipython()
if ipython is not None:
    ipython.run_line_magic("matplotlib", "inline")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

DAY_ORDER = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
MONTH_ORDER = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]


def nth_weekday_of_month(year: int, month: int, weekday: int, occurrence: int) -> pd.Timestamp:
    first_day = pd.Timestamp(year=year, month=month, day=1)
    day_offset = (weekday - first_day.weekday()) % 7
    return first_day + pd.Timedelta(days=day_offset + 7 * (occurrence - 1))


def pacific_dst_bounds_utc(year: int) -> tuple[pd.Timestamp, pd.Timestamp]:
    dst_start = nth_weekday_of_month(year, 3, 6, 2) + pd.Timedelta(hours=10)
    dst_end = nth_weekday_of_month(year, 11, 6, 1) + pd.Timedelta(hours=9)
    return dst_start.tz_localize("UTC"), dst_end.tz_localize("UTC")


def utc_to_pacific_naive(values: pd.Series) -> pd.Series:
    utc_values = pd.to_datetime(values, utc=True, errors="coerce")
    local_values = utc_values - pd.Timedelta(hours=8)

    valid_years = utc_values.dropna().dt.year.unique()
    for year in valid_years:
        start_utc, end_utc = pacific_dst_bounds_utc(int(year))
        dst_mask = (utc_values >= start_utc) & (utc_values < end_utc)
        local_values.loc[dst_mask] = utc_values.loc[dst_mask] - pd.Timedelta(hours=7)

    return local_values.dt.tz_localize(None)
```

```python
candidate_roots = [Path.cwd(), Path.cwd().parent]
PROJECT_ROOT = next(
    (path.resolve() for path in candidate_roots if (path / "data" / "processed").exists()),
    Path.cwd().resolve(),
)

DATA_DIR = PROJECT_ROOT / "data" / "processed"
FIGURES_DIR = PROJECT_ROOT / "reports" / "figures"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

session_path = DATA_DIR / "acn_sessions_flat.csv"
hourly_path = DATA_DIR / "hourly_demand.csv"

session_df = pd.read_csv(session_path)
hourly_df = pd.read_csv(hourly_path)

print(f"Project root: {PROJECT_ROOT}")
print(f"Session-level dataset shape: {session_df.shape}")
print(f"Hourly dataset shape: {hourly_df.shape}")
```

```python
def save_figure(filename: str) -> None:
    plt.savefig(FIGURES_DIR / filename, dpi=300, bbox_inches="tight")
    print(f"Saved figure: {filename}")


session_df["kWhDelivered"] = pd.to_numeric(session_df["kWhDelivered"], errors="coerce")
session_df["connection_utc"] = pd.to_datetime(session_df["connectionTime"], utc=True)
session_df["disconnect_utc"] = pd.to_datetime(session_df["disconnectTime"], utc=True)
session_df["done_charging_utc"] = pd.to_datetime(session_df["doneChargingTime"], utc=True)

session_df["connection_local"] = utc_to_pacific_naive(session_df["connectionTime"])
session_df["disconnect_local"] = utc_to_pacific_naive(session_df["disconnectTime"])
session_df["done_charging_local"] = utc_to_pacific_naive(session_df["doneChargingTime"])

session_df["duration_hours"] = (
    session_df["disconnect_utc"] - session_df["connection_utc"]
).dt.total_seconds() / 3600
session_df["active_charging_hours"] = (
    session_df["done_charging_utc"] - session_df["connection_utc"]
).dt.total_seconds() / 3600
session_df["local_date"] = session_df["connection_local"].dt.floor("D")
session_df["local_hour"] = session_df["connection_local"].dt.hour
session_df["weekday_name"] = pd.Categorical(
    session_df["connection_local"].dt.day_name(),
    categories=DAY_ORDER,
    ordered=True,
)
session_df["month_name"] = pd.Categorical(
    session_df["connection_local"].dt.month_name(),
    categories=MONTH_ORDER,
    ordered=True,
)

hourly_df["session_count"] = pd.to_numeric(hourly_df["session_count"], errors="coerce")
hourly_df["total_kwh"] = pd.to_numeric(hourly_df["total_kwh"], errors="coerce")
hourly_df["hour_utc"] = pd.to_datetime(hourly_df["hour"], utc=True)
hourly_df["hour_local"] = utc_to_pacific_naive(hourly_df["hour"])
hourly_df["local_date"] = hourly_df["hour_local"].dt.floor("D")
hourly_df["local_hour"] = hourly_df["hour_local"].dt.hour
hourly_df["weekday_name"] = pd.Categorical(
    hourly_df["hour_local"].dt.day_name(),
    categories=DAY_ORDER,
    ordered=True,
)
hourly_df["month_name"] = pd.Categorical(
    hourly_df["hour_local"].dt.month_name(),
    categories=MONTH_ORDER,
    ordered=True,
)
hourly_df["is_weekend_local"] = hourly_df["weekday_name"].isin(["Saturday", "Sunday"])

daily_summary = (
    session_df.groupby("local_date", as_index=False)
    .agg(session_count=("_id", "count"), total_kwh=("kWhDelivered", "sum"))
    .sort_values("local_date")
)
daily_summary["sessions_7d_avg"] = daily_summary["session_count"].rolling(7, min_periods=1).mean()
daily_summary["energy_7d_avg"] = daily_summary["total_kwh"].rolling(7, min_periods=1).mean()
daily_summary["weekday_name"] = pd.Categorical(
    daily_summary["local_date"].dt.day_name(),
    categories=DAY_ORDER,
    ordered=True,
)
daily_summary["month_name"] = pd.Categorical(
    daily_summary["local_date"].dt.month_name(),
    categories=MONTH_ORDER,
    ordered=True,
)
daily_summary["is_weekend"] = daily_summary["weekday_name"].isin(["Saturday", "Sunday"])
```

## 1. Dataset overview

The project uses two processed datasets:
- `acn_sessions_flat.csv`: session-level records with connection time, disconnection time, and delivered energy
- `hourly_demand.csv`: an hourly aggregated dataset for forecasting, including session count and total delivered energy

The table below provides a compact summary before plotting.


```python
overview = pd.DataFrame(
    [
        {
            "dataset": "Session-level",
            "rows": len(session_df),
            "columns": session_df.shape[1],
            "date_range": f"{session_df['connection_local'].min():%Y-%m-%d} to {session_df['connection_local'].max():%Y-%m-%d}",
        },
        {
            "dataset": "Hourly demand",
            "rows": len(hourly_df),
            "columns": hourly_df.shape[1],
            "date_range": f"{hourly_df['hour_local'].min():%Y-%m-%d} to {hourly_df['hour_local'].max():%Y-%m-%d}",
        },
    ]
)

audit_metrics = pd.DataFrame(
    {
        "metric": [
            "Total charging sessions",
            "Total delivered energy (kWh)",
            "Average energy per session (kWh)",
            "Median session duration (hours)",
            "Average hourly session count",
            "Average hourly delivered energy (kWh)",
        ],
        "value": [
            f"{len(session_df):,}",
            f"{session_df['kWhDelivered'].sum():,.1f}",
            f"{session_df['kWhDelivered'].mean():.2f}",
            f"{session_df['duration_hours'].median():.2f}",
            f"{hourly_df['session_count'].mean():.2f}",
            f"{hourly_df['total_kwh'].mean():.2f}",
        ],
    }
)

display(overview)
display(audit_metrics)
```

## 2. Daily charging demand trend

The next figure tracks both the number of sessions and the total energy delivered per day. A 7-day moving average is added to smooth short-term volatility and reveal the broader trend.


```python
fig, axes = plt.subplots(2, 1, figsize=(15, 10), sharex=True)

axes[0].plot(daily_summary["local_date"], daily_summary["session_count"], color="#4C78A8", alpha=0.4, label="Daily count")
axes[0].plot(daily_summary["local_date"], daily_summary["sessions_7d_avg"], color="#E45756", linewidth=2.5, label="7-day average")
axes[0].set_title("Daily Charging Sessions")
axes[0].set_xlabel("")
axes[0].set_ylabel("Sessions")
axes[0].legend()

axes[1].plot(daily_summary["local_date"], daily_summary["total_kwh"], color="#72B7B2", alpha=0.4, label="Daily energy")
axes[1].plot(daily_summary["local_date"], daily_summary["energy_7d_avg"], color="#F58518", linewidth=2.5, label="7-day average")
axes[1].set_title("Daily Delivered Energy")
axes[1].set_xlabel("Local date")
axes[1].set_ylabel("kWh")
axes[1].legend()

plt.suptitle("Daily EV Charging Activity in Local California Time", y=1.02)
plt.tight_layout()
save_figure("daily_usage_trend.png")
plt.show()
```

## 3. Intraday demand pattern

Because the source timestamps are in UTC, the hourly profile must be converted to local time before interpretation. The following figure shows average hourly session count and average hourly delivered energy in California local time.


```python
hour_profile = (
    hourly_df.groupby("local_hour", as_index=False)
    .agg(avg_session_count=("session_count", "mean"), avg_total_kwh=("total_kwh", "mean"))
)

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

axes[0].bar(hour_profile["local_hour"], hour_profile["avg_session_count"], color="#4C78A8")
axes[0].set_title("Average Hourly Session Count")
axes[0].set_xlabel("Hour of day")
axes[0].set_ylabel("Average sessions")
axes[0].set_xticks(range(24))

axes[1].bar(hour_profile["local_hour"], hour_profile["avg_total_kwh"], color="#F58518")
axes[1].set_title("Average Hourly Delivered Energy")
axes[1].set_xlabel("Hour of day")
axes[1].set_ylabel("Average kWh")
axes[1].set_xticks(range(24))

plt.tight_layout()
save_figure("hourly_profile_local_time.png")
plt.show()
```

## 4. Weekday and weekend behavior

The heatmap shows how hourly demand changes across the week, while the line chart compares the average hourly pattern between weekdays and weekends.


```python
weekday_hour_heatmap = (
    hourly_df.pivot_table(
        index="weekday_name",
        columns="local_hour",
        values="session_count",
        aggfunc="mean",`n        observed=False,`n    )
    .reindex(DAY_ORDER)
)

plt.figure(figsize=(16, 6))
plt.imshow(weekday_hour_heatmap, aspect="auto", cmap="YlGnBu")
plt.colorbar(label="Average sessions")
plt.title("Average Hourly Session Count by Weekday and Local Hour")
plt.xlabel("Hour of day")
plt.ylabel("")
plt.xticks(ticks=range(24), labels=range(24))
plt.yticks(ticks=range(len(DAY_ORDER)), labels=DAY_ORDER)
plt.tight_layout()
save_figure("weekday_hour_heatmap.png")
plt.show()

weekpart_profile = (
    hourly_df.groupby(["is_weekend_local", "local_hour"], as_index=False)["session_count"]
    .mean()
    .rename(columns={"session_count": "avg_session_count"})
)

weekday_profile = weekpart_profile.loc[~weekpart_profile["is_weekend_local"]]
weekend_profile = weekpart_profile.loc[weekpart_profile["is_weekend_local"]]

plt.figure(figsize=(12, 6))
plt.plot(weekday_profile["local_hour"], weekday_profile["avg_session_count"], marker="o", linewidth=2.5, label="Weekday")
plt.plot(weekend_profile["local_hour"], weekend_profile["avg_session_count"], marker="o", linewidth=2.5, label="Weekend")
plt.title("Weekday vs Weekend Charging Profile")
plt.xlabel("Hour of day")
plt.ylabel("Average sessions")
plt.xticks(range(24))
plt.legend()
plt.tight_layout()
save_figure("weekday_vs_weekend_profile.png")
plt.show()

```

## 5. Calendar-level demand summary

The bars below show average daily demand by weekday and by month. Using average daily demand is better than raw totals here because the dataset covers some partial months.


```python
weekday_daily_avg = (
    daily_summary.groupby("weekday_name", observed=False)["session_count"]
    .mean()
    .reindex(DAY_ORDER)
    .reset_index(name="avg_daily_sessions")
)

month_daily_avg = (
    daily_summary.groupby("month_name", observed=False)["session_count"]
    .mean()
    .reindex(MONTH_ORDER)
    .reset_index(name="avg_daily_sessions")
)

fig, axes = plt.subplots(1, 2, figsize=(20, 6))

axes[0].bar(weekday_daily_avg["weekday_name"], weekday_daily_avg["avg_daily_sessions"], color="#4C78A8")
axes[0].set_title("Average Daily Sessions by Weekday")
axes[0].set_xlabel("")
axes[0].set_ylabel("Average daily sessions")
axes[0].tick_params(axis="x", rotation=30)

axes[1].bar(month_daily_avg["month_name"], month_daily_avg["avg_daily_sessions"], color="#54A24B")
axes[1].set_title("Average Daily Sessions by Month")
axes[1].set_xlabel("")
axes[1].set_ylabel("Average daily sessions")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
save_figure("calendar_demand_summary.png")
plt.show()
```

## 6. Session-level charging behavior

The final figure summarizes the distribution of delivered energy, the distribution of session duration, and the relationship between session duration and delivered energy.


```python
plot_source = session_df[["duration_hours", "kWhDelivered"]].dropna()
plot_sample = plot_source.sample(n=min(4000, len(plot_source)), random_state=42)

fig, axes = plt.subplots(1, 3, figsize=(21, 6))

axes[0].hist(session_df["kWhDelivered"].dropna(), bins=40, color="#4C78A8", alpha=0.85)
axes[0].set_title("Delivered Energy Distribution")
axes[0].set_xlabel("kWh delivered")
axes[0].set_ylabel("Session count")

axes[1].hist(session_df["duration_hours"].dropna(), bins=40, color="#F58518", alpha=0.85)
axes[1].set_title("Session Duration Distribution")
axes[1].set_xlabel("Duration (hours)")
axes[1].set_ylabel("Session count")

axes[2].scatter(plot_sample["duration_hours"], plot_sample["kWhDelivered"], alpha=0.35, s=18, color="#54A24B")
axes[2].set_title("Duration vs Delivered Energy")
axes[2].set_xlabel("Duration (hours)")
axes[2].set_ylabel("kWh delivered")

plt.tight_layout()
save_figure("session_behavior_summary.png")
plt.show()
```
